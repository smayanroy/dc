```python
#photo to be used is "C:\Users\SMAYA\Desktop\dp.jpg"\
import cv2
import numpy as np
```

```python
dp= cv2.imread("C:\\Users\\SMAYA\\Desktop\\dp.jpg")
print(dp.shape)
dp_gray= cv2.cvtColor(dp, cv2.COLOR_BGR2GRAY)
cv2.imwrite("C:\\Users\\SMAYA\\Desktop\\dp_gray.jpg", dp_gray)
cv2.imshow("Gray Image", dp_gray)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
dp_flips= cv2.flip(dp, 1)
cv2.imshow("Flipped Image along 1", dp_flips)
cv2.waitKey(5000)
dp_flips = cv2.flip(dp, 0)
cv2.imshow("Flipped Image along 0", dp_flips)
cv2.waitKey(5000)
dp_flips = cv2.flip(dp, -1)
cv2.imshow("Flipped Image along -1", dp_flips)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
dp_crop1= dp[0:320, 0:320]
cv2.imshow("Cropped Image 1", dp_crop1)
cv2.waitKey(5000)
dp_crop2= dp[320:640, 320:640]
cv2.imshow("Cropped Image 2", dp_crop2)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
dp_neg = cv2.bitwise_not(dp)
cv2.imshow("Negative Image", dp_neg)
cv2.waitKey(5000)
dp_neg_gray = cv2.cvtColor(dp_neg, cv2.COLOR_BGR2GRAY)
cv2.imshow("Negative Gray Image", dp_neg_gray)
cv2.waitKey(5000)
dp_bright = dp + 20
cv2.imshow("Brightened Image", dp_bright)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```
