```python
import cv2

img = cv2.imread(r"C:\Users\SMAYA\Desktop\dp.jpg")

# Convert from BGR to LAB color space
lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
l, a, b = cv2.split(lab)

# Apply CLAHE to the L (Lightness) channel
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(1,1))
cl1 = clahe.apply(l)

# Merge channels back and convert to BGR
updated_lab = cv2.merge((cl1, a, b))
result = cv2.cvtColor(updated_lab, cv2.COLOR_LAB2BGR)

cv2.imwrite(r"C:\Users\SMAYA\Desktop\dp_clahe.jpg", result)
```

```python
import cv2
while True:
    cv2.imshow("Original Image", img)
    cv2.imshow("CLAHE Image", result)

    # Wait for the 'q' key to be pressed to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
```
